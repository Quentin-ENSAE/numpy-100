#### 29. How to round away from zero a float array ? (★☆☆)

````python
np.random.seed(0)
x = np.random.uniform(-10, 10, size=(1,10))
print('x avant modification : ', x)

x[x > 0] = np.ceil(x[x > 0])
x[x <= 0] = np.floor(x[x <= 0])

print('x après modification : ', x)
````
